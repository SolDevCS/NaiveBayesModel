```python
import numpy as np
# We used NumPy Library, a Python library containing the array data type to be used and numerical operations to be performed for our Naive Bayes model.
# In here, it is assigned the alias np to access the methods easily.

class NaiveBayes:
  # Demonstrated in the attached video, this is where we define a Naive Bayes classifier from scratch. This will house the
  # the methods for training the model and making predictions.
  
  def fit(self, X, y):
    # This method is defined to get the training samples and training labels and to train the model. It has parameters for the class priors or the commonness
    # of each class in the data and feature statistics,

    n_samples, n_features = X.shape
    # We extract the number of samples and number of features to know how many data points and features it is working with. It is assumed that 
    # what we get is a NumPy array with shape or the samples and features.

    self._classes = np.unique(y)
    # For gathering, it is important that each unique class is gathered to make our computations individually during training. In this 
    # code, we store the distinct class labels found in the training labels stored in y.

    n_classes = len(self._classes)
    # We get the total number of classes for the correct storage arrays for the next line. It is calculated by using the length method
    # in the variable we defined earlier that stores the unique classes.

    self._mean = np.zeros((n_classes, n_features), dtype=np.float64)
    # A NumPy array initialized with zeroes is assigned for the mean of each feature on every class during training using the total
    # number of classes and number of features for each combination. It is in the float64 data type format to make it clear that the model works with decimals.

    self._var = np.zeros((n_classes, n_features), dtype=np.float64)
    # We also initialize a NumPy array of zeroes with the same parameters to store the variance. The formula uses Gaussian probability formula
    # to compute the likeliness of a feature value given a class and each class must have a prior for computing the overall posterior or
    # the final predicted probability.

    self._priors = np.zeros(n_classes, dtype=np.float64)
    # The array for the priors is also initialized for calculating how common each class is before observing features. It will only
    # contain the total number of classes and the float64 format is used.

    for idx, c in enumerate(self._classes):
      # In here, we will loop over each class to compute for each class statistic. The recently created NumPy arrays like the mean, variance, and 
      # priors will be updated for the current class based on its samples.

      X_c = X[y == c]
      # The samples of the current class c need to be extracted. The samples in array X

      self._mean[idx, :] = X_c.mean(axis=0)
      self._var[idx, :] = X_c.var(axis=0)
      self._priors[idx] = X_c.shape[0] / float(n_samples)

  def predict(self, X):
      y_pred = [self._predict(x) for x in X]
      return np.array(y_pred)

  def _predict(self, x):
      posteriors = []

      # calculate posterior probability for each class
      for idx, c in enumerate(self._classes):
          prior = np.log(self._priors[idx])
          posterior = np.sum(np.log(self._pdf(idx, x)))
          posterior = posterior + prior
          posteriors.append(posterior)

      # return class with the highest posterior
      return self._classes[np.argmax(posteriors)]

  def _pdf(self, class_idx, x):
      mean = self._mean[class_idx]
      var = self._var[class_idx]
      numerator = np.exp(-((x - mean) ** 2) / (2 * var))
      denominator = np.sqrt(2 * np.pi * var)
      return numerator / denominator

# Testing
if __name__ == "__main__":
    # Imports
    from sklearn.model_selection import train_test_split
    from sklearn import datasets

    def accuracy(y_true, y_pred):
        accuracy = np.sum(y_true == y_pred) / len(y_true)
        return accuracy

    X, y = datasets.make_classification(
        n_samples=1000, n_features=10, n_classes=2, random_state=123
    )
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, random_state=123
    )

    nb = NaiveBayes()
    nb.fit(X_train, y_train)
    predictions = nb.predict(X_test)

    print("Naive Bayes classification accuracy", accuracy(y_test, predictions))
```
